**Problem**

Solve the following equation

$$\sqrt{x + \sqrt{x + \sqrt{x + \sqrt{...}}}} = x$$

</div>

<div style="font-family: 'Times New Roman', serif;">

The intuitive solution :

$$\sqrt{x + \sqrt{x + \sqrt{x + \sqrt{...}}}} = x$$
$$\to x + \underbrace{\sqrt{x + \sqrt{x + \sqrt{...}}}}_{x} = x^2 \to x+x = x^2$$
$$\therefore x = 0 \text{ or } x = 2$$
</div>

<div style="font-family: 'Times New Roman', serif;">


**However for the solution to work, the existence of the limit must be first proved.**

Example : $x + (x + (x + ...)) = 10$

With the same approach, $x + \underbrace{(x + (x + ...))}_{10} = 10 \to x + 10 = x \to  x = 0$ which is clearly a wrong answer.  
It must be first proven that given $a_{k+1} = x + a_k, \lim_{k\to\infty} a_k$ converges to a single point "S"

</div>



<div style="font-family: 'Times New Roman', serif;">

Let $a_1 = \sqrt{x}$ and $a_{n+1} = \sqrt{x + a_n}$

Then $$\sqrt{x + \sqrt{x + \sqrt{x + \sqrt{...}}}} = \lim_{n\to \infty}{a_n}$$

**Monotone Convergence Theorem**  


1) $a_n$ is bounded above :  

    Let $M = \text{max}(2,2x)$  
    At n = 1,  $a_1 = \sqrt{x} < M$  
    Assume $a_k < M$. Then, $a_{k+1} = \sqrt{x + a_k} < \sqrt{x+M}$   
    Since $x < \text{max}(2,2x) = M$, $a_{k+1} < \sqrt{x+M} < \sqrt{2M} \leq \sqrt{M^2} = M$
</br></br>    

2) $a_n$ is monotone increasing :  

    At n = 1,  $(a_1 = \sqrt{x}) \leq (a_2 = \sqrt{x + \sqrt{x}})$  
    Assume $a_k \leq a_{k+1}$, then $a_{k+1} = \sqrt{x+a_k} \leq \sqrt{x+a_{k+1}} = a_{k+2}$  


$\therefore$ $a_n$ is monotone increasing and bounded above, therefore by MCT:
$$0 \leq \lim_{n\to \infty}{a_n} = L(x) < \infty$$

</br>  

</div>

<div style="font-family: 'Times New Roman', serif;">

**Confirming the answer**

</div>

```python
import math

tolerance = 1e-10
max_iter = 1000

x = 2.0

A = math.sqrt(x)

for i in range(max_iter):

    A_next = math.sqrt(x + A)
    print(f"[{i}] A: {A_next}")

    if abs(A_next - A) < tolerance:
        break
    A = A_next
    
print("\nConverged value of A:", round(A_next,10))
```
